```python
# 01_D3
# geometri 2d opencv

black_img = np.zeros(shape=(512,512,3), dtype=np.int16)
cv.rectangle(black_img, pt1=(384,0), pt2=(510,150), color=(0,255,0), thickness=10)
cv.rectangle(black_img, pt1=(200,200), pt2=(300,300), color=(0,0,255), thickness=15)
cv.circle(black_img, center=(100,100), radius=50, color=(255,0,0), thickness=8)
cv.line(black_img, pt1=(0,0), pt2=(512,512), color=(255,255,255), thickness=5)
cv.circle(black_img, center=(400,350), radius=50, color=(0,255,0), thickness=-1)

font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img, text="hanzel", org=(10,500), fontFace=font, fontScale=4, color=(255,255,0), thickness=2, lineType=cv.LINE_AA)

plt.imshow(black_img)
```

```python
# 01_D3
# geometri int32 dtype
black_img2 = np.zeros(shape=(512,512,3), dtype=np.int32)
vertices = np.array([[100,300], [200,200], [400,300], [200,400]], dtype=np.int32)
print(f"vertices: {vertices}")

pts = vertices.reshape((-1,1,2))
print(f"pts: {pts}")

cv.polylines(black_img2, [pts], isClosed=True, color=(0,255,0), thickness=5)

plt.imshow(black_img2)
```

```python
# PERTANYAAN PRAKTIKUM
# nomor 1
imgX = cv.imread("/content/drive/MyDrive/colab_notebooks/gambar/ktp_ohim.jpg")
print(f"cv2_imshow = ", imgX[0,0])
cv2_imshow(imgX)
print(f"plt.imshow = ")
plt.imshow(imgX)

# jawaban:
# hasilnya berbeda karena opencv membaca gambar dalam urutan warna BGR, sedangkan
# plt.imshow() menanggap RGB. Akibatnya, warna merah dan biru tertukar pada plt.imshow()
```

```python
# PERTANYAAN PRAKTIKUM
# nomor 2
# jawaban = keduanya mempunyai hasil yang sama yaitu 0 pixel (hitam)
# yang membedakan hanyalah penggunaan memori, int16 menggunakan 16bit dan int32 membutuhkan 32bit
# int16 = nbytes = 1572864 | int32 = nbytes = 3145728

black_img16 = np.zeros(shape=(512,512,3), dtype=np.int16)
black_img32 = np.zeros(shape=(512,512,3), dtype=np.int32)
print(f"dtype = {black_img16.dtype}, min = {black_img16.min()}, max = {black_img16.max()}, nbytes = {black_img16.nbytes}")
print(f"dtype = {black_img32.dtype}, min = {black_img32.min()}, max = {black_img32.max()}, nbytes = {black_img32.nbytes}")

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("int16")
plt.imshow(black_img16)
plt.subplot(1,2,2)
plt.title("int32")
plt.imshow(black_img32)

```

```python
# PERTANYAAN PRAKTIKUM
# nomor 3
# jawaban = cv2_imshow() digunakan untuk menampilkan gambar di google colab, cv2.imshow() tidak bisa digunakan karena membutuhkan GUI window
# sedangkan cv2_imshow() -> google colab run melalui browser
```

```python
# PERTANYAAN PRAKTIKUM
# nomor 4
# jawaban = skimage.io.imread membaca gambardalam format rgb sehingga warna sesuai saat
# ditampilkan menggunakan plt.imshow()
# opencv membaca dalam format bgr, sehingga perlu cv2.cvtColor() agar menjadi rgb

img_skimage = io.imread("/content/drive/MyDrive/colab_notebooks/gambar/ktp_ohim.jpg")
print(f"shape = {img_skimage.shape}")
print(f"dtype = {img_skimage.dtype}")

img_cv = cv.imread("/content/drive/MyDrive/colab_notebooks/gambar/ktp_ohim.jpg")
img_rgb = cv.cvtColor(img_cv, cv.COLOR_BGR2RGB)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("skimage.io.imread")
plt.imshow(img_cv)
plt.subplot(1,2,2)
plt.title("cv2.cvtColor(bgr2rgb)")
plt.imshow(img_rgb)

# mana yang tertukar
y = 100
x = 100
print(f"posisi pixel: {y,x}")
print(f"skimage {img_skimage[y,x]}")
print(f"opencv bgr {img_cv[y,x]}")
print(f"opencv rgb {img_rgb[y,x]}")
```

```python
# TUGAS PRAKTIKUM - Penerapan
# nomor 1
img = cv.imread("/content/drive/MyDrive/colab_notebooks/gambar/ktp_ohim.jpg")
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

r = img_rgb[:,:,0]
g = img_rgb[:,:,1]
b = img_rgb[:,:,2]

red_blue = np.zeros_like(img_rgb)
red_blue[:,:,0] = r
red_blue[:,:,2] = b

green_blue = np.zeros_like(img_rgb)
green_blue[:,:,1] = g
green_blue[:,:,2] = b

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.title("red blue")
plt.imshow(red_blue)

plt.subplot(1,2,2)
plt.title("green blue")
plt.imshow(green_blue)
plt.show()
```

```python
# TUGAS PRAKTIKUM - Penerapan
# nomor 2
img = cv.imread("/content/drive/MyDrive/colab_notebooks/gambar/ktp_ohim.jpg")
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

plt.figure(figsize=(15,5))
plt.subplot(1,3,1)
plt.title("vertical")
plt.imshow(cv.flip(img_rgb, 0))

plt.subplot(1,3,2)
plt.title("horizontal")
plt.imshow(cv.flip(img_rgb, 1))

plt.subplot(1,3,3)
plt.title("vertical + horizontal")
plt.imshow(cv.flip(img_rgb, -1))
```

```python
# TUGAS PRAKTIKUM - Analisa
img_ohim = cv.imread("/content/drive/MyDrive/colab_notebooks/gambar/ktp_ohim.jpg")
gray = cv.cvtColor(img_ohim, cv.COLOR_BGR2GRAY)
cv2_imshow(gray)

sum_dark = np.sum(gray < 100)
sum_pixel = gray.size

percentage = (sum_dark / sum_pixel) * 100

print(f"pixel gelap : {sum_dark} | jumlah pixel : {sum_pixel} | persentase : {percentage} %")
```
