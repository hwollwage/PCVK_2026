```python
# 01_D2
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from PIL import Image
import matplotlib.pyplot as plt
```

```python
# 01_D2
urls = [
    "https://fastly.picsum.photos/id/29/4000/2670.jpg?hmac=rCbRAl24FzrSzwlR5tL-Aqzyu5tX_PA95VJtnUXegGU",
    "https://fastly.picsum.photos/id/28/4928/3264.jpg?hmac=GnYF-RnBUg44PFfU5pcw_Qs0ReOyStdnZ8MtQWJqTfA",
    "https://fastly.picsum.photos/id/18/2500/1667.jpg?hmac=JR0Z_jRs9rssQHZJ4b7xKF82kOj8-4Ackq75D_9Wmz8"
]

for url in urls:
  img = io.imread(url)
  img = cv.resize(img, (0,0), fx=0.5, fy=0.5)
  img_2 = cv.cvtColor(img, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((img, img_2))

  cv2_imshow(final_frame)
  print('\n')
```

```python
tinggi = img_2.shape[0]
lebar = img_2.shape[1]
```

```python
print(f"resolusi img: p * l -> {tinggi} * {lebar}")
cv2_imshow(img_2)
```

```python
# 01_D2

img_2 = cv.cvtColor(img, cv.COLOR_BGR2RGB)
img_3 = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# horizontal line
for y in range(lebar):
  img_3[int(tinggi / 2), y] = [255,255,255]

final_frame = cv.hconcat((img_2, img_3))
cv2_imshow(final_frame)
```

```python
# TUGAS PRAKTIKUM D2 (01_D2)
# garis vertikal
img_4 = cv.cvtColor(img, cv.COLOR_BGR2RGB)

for x in range(tinggi):
  img_4[x, int(lebar / 2)] = [255,255,255]

final_frame = cv.hconcat((img_2, img_4))
cv2_imshow(final_frame)
```

```python
# TUGAS PRAKTIKUM D2 (01_D2)
# garis menyilang diagonal

img_5 = cv.cvtColor(img, cv.COLOR_BGR2RGB)

cv.line(img, (0, 0), (lebar, tinggi), (255, 255, 255), 2)
cv.line(img, (lebar, 0), (0, tinggi), (255, 255, 255), 2)

diagonal = cv.hconcat((img_2, img_5))
cv2_imshow(diagonal)
```

```python
# TUGAS PRAKTIKUM D2 (01_D2)
# garis horizontal di tengah gambar

img_6 = cv.cvtColor(img, cv.COLOR_BGR2RGB)

tengah_x = int(lebar // 2)
tengah_y = int(tinggi // 2)

cv.line(img_6, (300, tengah_y), (700, tengah_y), (255,255,255), 5)

center_hor = cv.hconcat((img_2, img_6))
cv2_imshow(center_hor)
```

```python
# TUGAS PRAKTIKUM D2 (01_D2)
# kotak putih (loop)

img_7 = cv.cvtColor(img, cv.COLOR_BGR2RGB)

tengah_x = int(lebar // 2)
tengah_y = int(tinggi // 2)
for line in range(400):
  y = tengah_y - 200 + line
  cv.line(img_7, (300, y), (700, y), (255,255,255), 5)

kotak = cv.hconcat((img_2, img_7))
cv2_imshow(kotak)
```

```python
# TUGAS PRAKTIKUM D2 (01_D2)
# kotak putih (cv rect)

img_8 = cv.cvtColor(img, cv.COLOR_BGR2RGB)

tengah_x = int(lebar // 2)
tengah_y = int(tinggi // 2)

cv.rectangle(img_8, (tengah_x, 400), (1000, 700), (255,255,255), -1)

cvrect = cv.hconcat((img_2, img_8))
cv2_imshow(cvrect)
```
