```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
def dilasi_citra(F,w):
  #size image
  p,q= F.shape
  imgD= np.zeros((p,q), dtype=np.uint8)
  #Generate structure element dilasi (SED)
  SED= np.ones((w,w), dtype=np.uint8)
  constant1= (w-1)//2

  #Proses Dilasi
  for i in range(constant1, p-constant1):
    for j in range(constant1,q-constant1):
      temp= F[i-constant1:i+constant1+1, j-constant1:j+constant1+1]
      product= temp*SED
      imgD[i,j]= np.max(product)
  return imgD

img_d= cv2.imread('/content/drive/MyDrive/Colab Notebooks/gambar/plat.jpg',0)
ret, thresh = cv2.threshold(img_d, 125, 200, cv2.THRESH_BINARY)
w=5
imgD = dilasi_citra(thresh,w)

plt.subplot(131),plt.imshow(img_d,cmap = 'gray')
plt.title('Citra Awal'), plt.xticks([]), plt.yticks([])
plt.subplot(132),plt.imshow(imgD,cmap = 'gray')
plt.title('Citra Hasil Dilasi'), plt.xticks([]), plt.yticks([])
plt.show()
```

```python
import cv2 as cv
from matplotlib import pyplot as plt
from skimage import filters
from skimage.morphology import skeletonize

# read image
img = cv.imread('/content/drive/MyDrive/Colab Notebooks/gambar/lily.jpg',0)

binary = img > filters.threshold_triangle(img)

# true false to one
binary_cp = binary.copy()
binary_cp[binary_cp == True] = 1
binary_cp[binary_cp == False] = 0

# skeletonize image
skeleton = skeletonize(binary_cp)

# print images
plt.figure(figsize=(20,20))

plt.subplot(121),plt.imshow(img,cmap = 'gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])

plt.subplot(122),plt.imshow(skeleton,cmap = 'gray')
plt.title('skeleton Image'), plt.xticks([]), plt.yticks([])

plt.savefig("binary and bin_skeleton.png")
plt.show()
```

```python
from skimage import io
from matplotlib import pyplot as plt
from skimage import filters
from skimage.morphology import skeletonize

# read image
img = io.imread('/content/drive/MyDrive/Colab Notebooks/gambar/lily.jpg')[..., 0]

# Note: we want the black bits to be True, so use <
binary = img < filters.threshold_triangle(img)

# skeletonize image
skeleton = skeletonize(binary)

# print images
fig, ax = plt.subplots(1, 2)

ax[0].imshow(img, cmap='gray')
ax[0].set_title('original')
ax[0].set_axis_off()

# note the reversed colormap, gray_r
ax[1].imshow(skeleton, cmap='gray_r')
ax[1].set_title('skeleton (inverse)')
ax[1].set_axis_off()
plt.show()
```

```python
import os
import cv2
import matplotlib.pyplot as plt
import numpy as np
from google.colab import drive

drive.mount('/content/drive')

paths = ['/content/drive/MyDrive/Colab Notebooks/gambar/ktp1.jpg']

img_path = None
for p in paths:
  if os.path.exists(p):
    img_path = p
    break

if img_path is None:
  raise FileNotFoundError(
      'File KTP tidak ditemukan!'
  )

print(f'Memuat citra dari: {img_path}')
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

# 3. Adaptive Thresholding
thresh = cv2.adaptiveThreshold(
    img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY_INV, 11, 11
)

# 4. Operasi Morfologi
kernel_close = cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2))
kernel_open = cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2))

closed = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel_close)
cleaned = cv2.morphologyEx(closed, cv2.MORPH_OPEN, kernel_open)

plt.figure(figsize=(18, 6))

plt.subplot(141)
plt.imshow(img, cmap='gray')
plt.title('1. Citra KTP Asli')
plt.axis('off')

plt.subplot(142)
plt.imshow(thresh, cmap='gray')
plt.title('2. Adaptive Thresholding (Biner)')
plt.axis('off')

plt.subplot(143)
plt.imshow(closed, cmap='gray')
plt.title('3. Setelah Closing (Sambung Huruf)')
plt.axis('off')

plt.subplot(144)
plt.imshow(cleaned, cmap='gray')
plt.title('4. Hasil Akhir (+ Opening Bersih)')
plt.axis('off')

plt.tight_layout()
plt.show()
```
